```python
# Example notebook to run baseline + attention GNN

import matplotlib.pyplot as plt
from src.datasets import DIV2KSubset
from src.graph_utils import patch_to_graph
from src.models.ipg_baseline import IPGBaseline
from src.models.gnn_attention import GNNAttention
import torch

# Load dataset (subset)
dataset = DIV2KSubset(hr_dir='data/DIV2K_sub', lr_dir='data/DIV2K_sub_LR', patch_size=16)
lr_patches, hr_patches = dataset[0]

# Example: convert patches to graph
data = patch_to_graph(lr_patches)

# Baseline model
model = IPGBaseline(in_channels=16*16, hidden_channels=64, out_channels=16*16)
output = model(data)

# Reshape and visualize first patch
output_patch = output[0].detach().numpy().reshape(16,16)
plt.imshow(output_patch, cmap='gray')
plt.title("Baseline Patch Reconstruction")
plt.show()
```
