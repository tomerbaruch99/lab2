```python
import pandas as pd
import matplotlib.pyplot as plt

SCORED_CSV = "./evaluation/haifa_rag_strategies_scored.csv"

df = pd.read_csv(SCORED_CSV, encoding="utf-8")
print("Rows:", len(df))
df.head()
```

```python
# Summary by strategy

metrics = ["correctness", "faithfulness", "completeness", "conciseness", "overall"]
summary = df.groupby("strategy")[metrics].mean().reset_index()
summary
```

```python
# Overall score barplot

plt.figure(figsize=(10, 5))
plt.bar(summary["strategy"], summary["overall"])
plt.title("Overall LLM-Judge Score by Strategy")
plt.ylabel("Score (0-1)")
plt.xlabel("Strategy")
plt.ylim(0, 1)
plt.show()
```

```python
# All metrics by strategy

summary.set_index("strategy")[metrics].plot(kind="bar", figsize=(12, 6))
plt.title("Metrics by Strategy (Gemini Judge)")
plt.ylim(0, 1)
plt.ylabel("Score (0-1)")
plt.xticks(rotation=0)
plt.show()
```

```python
# Optional - per-question comparison table

pivot_overall = df.pivot_table(
    index="question",
    columns="strategy",
    values="overall"
)
pivot_overall.head()
```
